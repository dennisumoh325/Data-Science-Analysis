## Finding the movies with more than one genre

```python
#adding a column to the movie dataset for the number of genre each movie have
movie["Number of Genres"] = movie.loc[:, genres].sum(axis=1)
```

```python
movie.head()
```

```python
#this will subset the data as per the given condition
movie[movie['Number of Genres']>1]
```

### Business Insight

```python
df_merge = movie.merge(ratings, on = 'movie id', how = 'inner')
df_merge.head()
```

```python
#To find the average rating for each genre, we can use a for loop to iterate through the genres list and calculate the mean rating for each genre using the df_merge dataframe. The code below will print the average rating for each genre.
for i in genres:
    print(i,':' , df_merge[df_merge[i]==1].rating.mean())
```

- *'Film-Noir' genre is having the highest average ratings with an average rating of 3.92 followed by 'War' genre with rating of 3.81*
- *'Fantasy' genre is having the lowest average ratings with an average rating of 3.21*
- *Among 18 genres, only 5 genres have received a rating below 3.5 which means around 72% of genres are rated above average*

```python
avg_rating[avg_rating['Avg_rating']==5] #it will subset the data as per the specified condition
```

### Finding movies rated most of the time

```python
movie_count = df_merge.groupby(['movie title'])['rating'].count().reset_index()
movie_count
```

```python
movie_count.rename(columns={'rating':'Rating_counts'},inplace=True)
movie_count.head()
```

```python
movie_count.sort_values(ascending=False, by= 'Rating_counts')
```

- **Star Wars** has been rated most number of times (**583 times**) and there are few movies which have least rating counts like **Paris Was a Woman	,Á köldum klaka (Cold Fever)**.*

```python
## Checking for the movies which have more than 100 ratings. This will help us to find the movies which are rated by more than 100 users.
movie_100 = movie_count[movie_count['Rating_counts']>100] #this will subset the dataset 'movie_count' as per the specified condition
movie_100
```
